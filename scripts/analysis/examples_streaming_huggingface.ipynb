# Stream from Hugging Face

This notebook does not download the snapshot. `load_dataset(..., streaming=True)` reads parquet from the Hub on your machine as you iterate.

Do **not** stream `agent_session_logs` this way. Every row (including nested log payloads) is downloaded and decoded here. That is slow and repeats on every run. Snapshot the dataset and use [`examples_streaming.ipynb`](examples_streaming.ipynb) for logs.

# Setup

```python
from datasets import load_dataset
from tqdm.auto import tqdm

from agentlogs.schema import assert_version

REVISION = "v0.2"
assert_version(REVISION)
```

# Repositories
Warning: the schema of `repositories.languages` is currently not supported by HuggingFace. See the other example scripts for working analysis code for repositories. This will be fixed in v1.0.

```python
n_repositories = 0
n_repositories_with_tasks = 0
for row in tqdm(
    load_dataset("risenlab/agentlogs", "repositories", split="train", streaming=True, revision=REVISION),
    desc="repositories",
):
    n_repositories += 1
    if row["agent_tasks"]:
        n_repositories_with_tasks += 1

print(f"# repositories: {n_repositories}")
print(f"# repositories with agent tasks: {n_repositories_with_tasks}")
print(f"% repositories with agent tasks: {100.0 * n_repositories_with_tasks / n_repositories:.2f}%")
```

# Agent tasks

```python
n_tasks = 0
n_tasks_found = 0
n_tasks_with_sessions = 0
for row in tqdm(
    load_dataset("risenlab/agentlogs", "agent_tasks", split="train", streaming=True, revision=REVISION),
    desc="tasks",
):
    n_tasks += 1
    if row["found"]:
        n_tasks_found += 1
    if row["sessions"]:
        n_tasks_with_sessions += 1

print(f"# tasks: {n_tasks}")
print(f"# tasks found: {n_tasks_found}")
print(f"% tasks found: {100.0 * n_tasks_found / n_tasks:.2f}%")
print(f"# tasks with agent sessions: {n_tasks_with_sessions}")
print(f"% tasks with agent sessions: {100.0 * n_tasks_with_sessions / n_tasks_found:.2f}%")
```

# Agent sessions

```python
n_sessions = 0
n_logs_found = 0
for row in tqdm(
    load_dataset("risenlab/agentlogs", "agent_sessions", split="train", streaming=True, revision=REVISION),
    desc="sessions",
):
    n_sessions += 1
    if row["log_found"]:
        n_logs_found += 1

print(f"# sessions: {n_sessions}")
print(f"# sessions with logs found: {n_logs_found}")
print(f"% sessions with logs found: {100.0 * n_logs_found / n_sessions:.2f}%")
```

# Users

```python
n_users = 0
n_users_found = 0
for row in tqdm(
    load_dataset("risenlab/agentlogs", "users", split="train", streaming=True, revision=REVISION),
    desc="users",
):
    n_users += 1
    if row["found"]:
        n_users_found += 1

print(f"# users: {n_users}")
print(f"# users found: {n_users_found}")
print(f"% users found: {100.0 * n_users_found / n_users:.2f}%")
```
